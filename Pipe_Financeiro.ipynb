```python
import pandas as pd
import numpy as np
import os
```

```python
#Carregar arquivos financeiros
```

```python
pasta = r"D:\PROJETOS\Hamburgueria_BlackBull\02_Financeiro"
```

```python
#Carregar dados
```

```python
planilhas = [os.path.join(pasta, arquivo)for arquivo in os.listdir(pasta)]
```

```python
#Criar planilha consolidada de dados
```

```python
consol_fin = pd.DataFrame()
```

```python
for arquivo in planilhas:
    df = pd.read_excel(arquivo)
    consol_fin = pd.concat([consol_fin, df])
```
